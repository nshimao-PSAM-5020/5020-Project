## Milestone 01

### Idea 1: Hawaiian Pidgin --> Language Variation and Formality

My first idea stemmed from my interest in exploring Hawaiian Pidgin and how its expressions can communicate meanings beyond their direct English translations. For this machine learning project, I initially considered focusing exclusively on Hawaiian Pidgin, but I am concerned that the available dataset may be too small. My initial thought was to see whether semantic or emotional relationships between Hawaiian Pidgin words and phrases and their Standard English equivalents could be identified. As I thought more about this direction, I became interested in expanding the project to explore formal and informal language more broadly. Specifically, I would like to see whether machine learning can distinguish between formal and informal language while identifying expressions that communicate similar meanings across different levels of formality. Hawaiian Pidgin could potentially serve as an additional point of comparison within this broader exploration. I am still unsure how realistic this idea is as I could not find a specific dataset that seemed to be the right fit yet, but can continue to research and definitely thought about combining some or compiling my own.

### Idea 2: Language in Films or Sports? 

For my second idea, I kind of rode the wave of language, building on my first idea, but was torn between exploring movie/TV scripts or going more towards sports. One dataset I came across was the Cornell Movie-Dialogs Corpus, which contains conversations between characters from hundreds of movies. I was particularly interested in the interaction aspect of the dataset and exploring whether machine learning could identify patterns in how characters communicate with one another. Since the dataset includes information about who is speaking and the replies to others, I thought it could be interesting to see whether dialogue can reveal different interaction styles, as well as if these patterns vary across different character relationships or movie genres.

On the other hand, as I mentioned earlier, I was also interested in exploring language within sports, specifically how athletes and their performances are discussed through sports media and commentary. I thought it could be interesting to see if patterns in how athletes are described and whether the language used to discuss their performances relates to actual game statistics. I guess this is more like 2 ideas than one, but the brainstorming kind of came at the same time so I added them together here.

### Idea 3: NYC Datasets

My last idea was rooted in  wanting to explore something more visual, especially after working with images in class. I also have a strong personal interest in fashion, especially coming from Hawaii to NYC. One thought I had was to work with images of people's outfits around NYC and explore patterns in how people dress. This also led me to explore some of the NYC Open Data datasets, particularly the NYC Businesses dataset, the Storefronts Reported Vacant or Not dataset, and PLUTO, which provides information about NYC properties and buildings. My initial thought was to focus specifically on fashion-related businesses, but I am also open to expanding to NYC retail businesses more broadly if that provides more opportunities for combining datasets and exploring different modeling approaches. One concern I have is that fashion in NYC may be an overexplored topic, so I would want to find a more unique angle. I also recognize that the idea may need to be narrowed down depending on what the data can support, but I wanted to share it as a starting point because I think there could be some interesting possibilities.
